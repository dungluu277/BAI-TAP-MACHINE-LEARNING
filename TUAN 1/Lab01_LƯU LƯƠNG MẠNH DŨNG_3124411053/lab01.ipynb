## Bài 8. Autograd cơ bản

```python
import torch

x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

```python

dy_dx_thu_cong = 2 * 3.6
print("Dao ham thu cong (2x tai x=3.6):", dy_dx_thu_cong)
print("x.grad (tu autograd):", x.grad.item())
```

```python

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()

print("x2 =", x2)
print("y2 =", y2)
print("x2.grad =", x2.grad)

dy2_dx_thu_cong = 6 * 3.6 + 2
print("Dao ham thu cong (6x+2 tai x=3.6):", dy2_dx_thu_cong)
```

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

np.random.seed(0)
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

```python

print("He so mong muon: [1, 2, 3]")
print("He so hoc duoc:")
print(w.detach().numpy().ravel())
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

random.seed(0)
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python

print("A + B =", (A + B).item(), " (mong muon 9)")
print("C - D =", (C - D).item(), " (mong muon 1)")
print("A + C =", (A + C).item(), " (mong muon 8)")
print("B - D =", (B - D).item(), " (mong muon 2)")
```

```python

import numpy as np
import torch

np.random.seed(0)
polynomial2 = np.poly1d([2, -3, 5])
N = 20

X2 = np.random.randn(N, 1) * 5
Y2 = polynomial2(X2)

XX2 = np.hstack([
    X2 * X2,
    X2,
    np.ones_like(X2)
])

w2 = torch.randn(3, 1, requires_grad=True)
x2t = torch.tensor(XX2, dtype=torch.float32)
y2t = torch.tensor(Y2, dtype=torch.float32)

optimizer2 = torch.optim.NAdam([w2], lr=0.01)

print("Initial coefficients:")
print(w2)

for epoch in range(1000):
    optimizer2.zero_grad()
    y_pred2 = x2t @ w2
    mse2 = torch.mean(torch.square(y2t - y_pred2))
    mse2.backward()
    optimizer2.step()

print("Learned coefficients:")
print(w2)
print("\nHe so mong muon: [2, -3, 5]")
print("He so hoc duoc:", w2.detach().numpy().ravel())
```
