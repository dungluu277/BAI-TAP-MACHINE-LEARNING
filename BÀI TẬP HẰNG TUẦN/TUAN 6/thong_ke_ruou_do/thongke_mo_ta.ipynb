## 1. Nạp dữ liệu và tính thống kê mô tả

```python
import pandas as pd

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)

df = pd.read_csv("winequality-red.csv")

print("Kich thuoc du lieu:", df.shape)
print("\nKieu du lieu:\n", df.dtypes)
print("\nGia tri thieu:\n", df.isnull().sum())
print("So dong trung lap:", df.duplicated().sum())

print("\n5 dong dau:\n", df.head())

mo_ta = df.describe().T
mo_ta["median"] = df.median()
mo_ta["var"] = df.var()
mo_ta["skew"] = df.skew()
mo_ta["kurtosis"] = df.kurtosis()
mo_ta["range"] = df.max() - df.min()
mo_ta["IQR"] = df.quantile(0.75) - df.quantile(0.25)
print("\nThong ke mo ta:\n", mo_ta.round(4))
mo_ta.round(4).to_csv("thong_ke_mo_ta.csv")
```

## 2. Phân tích chất lượng rượu, tương quan và giá trị ngoại lệ

```python
print("\nMode:\n", df.mode().iloc[0])

dem = df["quality"].value_counts().sort_index()
ty_le = (df["quality"].value_counts(normalize=True).sort_index() * 100).round(2)
print("\nPhan bo chat luong ruou:\n", pd.DataFrame({"so_luong": dem, "ty_le_%": ty_le}))

theo_chat_luong = df.groupby("quality").agg(["mean", "std"]).round(4)
print("\nThong ke theo nhom chat luong:\n", theo_chat_luong)
theo_chat_luong.to_csv("thong_ke_theo_chat_luong.csv")

tuong_quan = df.corr().round(4)
print("\nMa tran tuong quan:\n", tuong_quan)
tuong_quan.to_csv("ma_tran_tuong_quan.csv")
print("\nTuong quan voi quality:\n", tuong_quan["quality"].drop("quality").sort_values(ascending=False))

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
outliers = ((df < Q1 - 1.5 * IQR) | (df > Q3 + 1.5 * IQR)).sum()
print("\nSo luong ngoai le (IQR) theo thuoc tinh:\n", outliers)
```
