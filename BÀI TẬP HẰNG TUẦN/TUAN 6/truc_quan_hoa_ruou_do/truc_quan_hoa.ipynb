## 1. Phân bố chất lượng rượu đỏ

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_csv("winequality-red.csv")
features = [c for c in df.columns if c != "quality"]

plt.figure(figsize=(8, 5))
ax = sns.countplot(x="quality", data=df, palette="Reds")
for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + p.get_width() / 2, p.get_height()), ha="center", va="bottom")
plt.title("Phan bo chat luong ruou do")
plt.xlabel("Chat luong")
plt.ylabel("So luong")
plt.tight_layout()
plt.savefig("01_phan_bo_chat_luong.png", dpi=150)
plt.show()
```

## 2. Tỷ lệ các mức chất lượng

```python
plt.figure(figsize=(8, 8))
counts = df["quality"].value_counts().sort_index()
plt.pie(counts, labels=counts.index, autopct="%1.1f%%", colors=sns.color_palette("Reds", len(counts)))
plt.title("Ty le cac muc chat luong")
plt.savefig("02_ty_le_chat_luong.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3. Phân bố của các thuộc tính

```python
fig, axes = plt.subplots(3, 4, figsize=(18, 11))
for ax, col in zip(axes.ravel(), features):
    sns.histplot(df[col], kde=True, ax=ax, color="darkred")
    ax.set_title(col)
axes.ravel()[-1].axis("off")
plt.suptitle("Phan bo cua cac thuoc tinh", fontsize=16)
plt.tight_layout()
plt.savefig("03_histogram_thuoc_tinh.png", dpi=150)
plt.show()
```

## 4. Boxplot các thuộc tính

```python
fig, axes = plt.subplots(3, 4, figsize=(18, 11))
for ax, col in zip(axes.ravel(), features):
    sns.boxplot(y=df[col], ax=ax, color="salmon")
    ax.set_title(col)
axes.ravel()[-1].axis("off")
plt.suptitle("Boxplot cac thuoc tinh", fontsize=16)
plt.tight_layout()
plt.savefig("04_boxplot_thuoc_tinh.png", dpi=150)
plt.show()
```

## 5. Boxplot thuộc tính theo chất lượng

```python
fig, axes = plt.subplots(3, 4, figsize=(18, 11))
for ax, col in zip(axes.ravel(), features):
    sns.boxplot(x="quality", y=col, data=df, ax=ax, palette="Reds")
    ax.set_title(f"{col} theo chat luong")
axes.ravel()[-1].axis("off")
plt.tight_layout()
plt.savefig("05_boxplot_theo_chat_luong.png", dpi=150)
plt.show()
```

## 6. Heatmap ma trận tương quan

```python
plt.figure(figsize=(12, 9))
corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Ma tran tuong quan")
plt.tight_layout()
plt.savefig("06_heatmap_tuong_quan.png", dpi=150)
plt.show()
```

## 7. Tương quan của các thuộc tính với chất lượng

```python
plt.figure(figsize=(8, 6))
corr["quality"].drop("quality").sort_values().plot(kind="barh", color="firebrick")
plt.title("Tuong quan cac thuoc tinh voi chat luong")
plt.tight_layout()
plt.savefig("07_tuong_quan_voi_chat_luong.png", dpi=150)
plt.show()
```

## 8. Scatter plot các cặp thuộc tính

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))
sns.scatterplot(x="alcohol", y="volatile acidity", hue="quality", data=df, palette="RdYlGn", ax=axes[0, 0])
sns.scatterplot(x="alcohol", y="density", hue="quality", data=df, palette="RdYlGn", ax=axes[0, 1])
sns.scatterplot(x="fixed acidity", y="pH", hue="quality", data=df, palette="RdYlGn", ax=axes[1, 0])
sns.scatterplot(x="free sulfur dioxide", y="total sulfur dioxide", hue="quality", data=df, palette="RdYlGn", ax=axes[1, 1])
plt.tight_layout()
plt.savefig("08_scatter_cac_cap_thuoc_tinh.png", dpi=150)
plt.show()
```

## 9. Phân bố nồng độ cồn theo chất lượng

```python
plt.figure(figsize=(9, 6))
sns.violinplot(x="quality", y="alcohol", data=df, palette="Reds")
plt.title("Nong do con theo chat luong")
plt.tight_layout()
plt.savefig("09_violin_alcohol_chat_luong.png", dpi=150)
plt.show()
```

## 10. Pairplot các thuộc tính chính

```python
top = corr["quality"].drop("quality").abs().sort_values(ascending=False).head(4).index.tolist() + ["quality"]
g = sns.pairplot(df[top], hue="quality", palette="RdYlGn", corner=True, plot_kws={"s": 12})
g.savefig("10_pairplot_thuoc_tinh_chinh.png", dpi=120)
plt.show()
```

## 11. Giá trị trung bình các thuộc tính theo chất lượng

```python
plt.figure(figsize=(9, 6))
df.groupby("quality")[["alcohol", "sulphates", "volatile acidity", "citric acid"]].mean().plot(marker="o", figsize=(9, 6))
plt.title("Gia tri trung binh cac thuoc tinh chinh theo chat luong")
plt.xlabel("Chat luong")
plt.tight_layout()
plt.savefig("11_trung_binh_theo_chat_luong.png", dpi=150)
plt.show()
```
